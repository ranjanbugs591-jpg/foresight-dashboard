```python
sales["Date"] = pd.to_datetime(sales["Date"])
calendar["date"] = pd.to_datetime(calendar["date"])
inventory["Snapshot_Date"] = pd.to_datetime(inventory["Snapshot_Date"])
sku["Launch_Date"] = pd.to_datetime(sku["Launch_Date"])

print(sales["Date"].dtype, calendar["date"].dtype, inventory["Snapshot_Date"].dtype, sku["Launch_Date"].dtype)
```

```python
weekly = (
    sales
    .groupby(["SKU", pd.Grouper(key="Date", freq="W-MON")])
    .agg(
        Units_Sold=("Units_Sold", "sum"),
        Revenue=("Revenue", "sum"),
        Price=("Price", "mean"),
        Promotion=("Promotion", "max")
    )
    .reset_index()
)

print(weekly.shape)
weekly.head()
```

```python
weekly = weekly.merge(sku, on="SKU", how="left")

print(weekly.shape)
weekly.head()
```

```python
print(calendar[["season", "holiday", "is_holiday", "promotion_event"]].head(10))
print(calendar["season"].unique())
print(calendar["holiday"].unique())
print(calendar["promotion_event"].unique())
```

```python
calendar["is_promo_event"] = calendar["promotion_event"].notna().astype(int)

calendar_weekly = (
    calendar
    .groupby(pd.Grouper(key="date", freq="W-MON"))
    .agg(
        is_holiday=("is_holiday", "max"),
        season=("season", "first"),
        is_promo_event=("is_promo_event", "max")
    )
    .reset_index()
)

print(calendar_weekly.shape)
calendar_weekly.head()
```

```python
weekly = weekly.merge(calendar_weekly, left_on="Date", right_on="date", how="left")
weekly = weekly.drop(columns=["date"])

print(weekly.shape)
weekly.head()
```

```python
weekly = weekly.sort_values(["SKU", "Date"])
inventory = inventory.sort_values(["SKU", "Snapshot_Date"])

weekly = pd.merge_asof(
    weekly,
    inventory,
    left_on="Date",
    right_on="Snapshot_Date",
    by="SKU",
    direction="backward"
)

print(weekly.shape)
weekly.isnull().sum()
```

```python
weekly = weekly.sort_values("Date")
inventory = inventory.sort_values("Snapshot_Date")

weekly = pd.merge_asof(
    weekly,
    inventory,
    left_on="Date",
    right_on="Snapshot_Date",
    by="SKU",
    direction="backward"
)

print(weekly.shape)
weekly.isnull().sum()
```

```python
weekly["snapshot_gap_days"] = (weekly["Date"] - weekly["Snapshot_Date"]).dt.days
print(weekly["snapshot_gap_days"].describe())
```

```python
weekly = weekly.sort_values(["SKU", "Date"]).reset_index(drop=True)

weekly["units_lag_1"] = weekly.groupby("SKU")["Units_Sold"].shift(1)
weekly["units_lag_2"] = weekly.groupby("SKU")["Units_Sold"].shift(2)
weekly["units_lag_4"] = weekly.groupby("SKU")["Units_Sold"].shift(4)

print(weekly[["SKU", "Date", "Units_Sold", "units_lag_1", "units_lag_2", "units_lag_4"]].head(8))
```

```python
weekly["units_roll_mean_4"] = (
    weekly.groupby("SKU")["Units_Sold"]
    .transform(lambda s: s.shift(1).rolling(4).mean())
)

print(weekly[["SKU", "Date", "Units_Sold", "units_lag_1", "units_roll_mean_4"]].head(8))
```

```python
weekly_model = weekly.dropna(
    subset=["units_lag_1", "units_lag_2", "units_lag_4", "units_roll_mean_4"]
).copy()

weekly_model = pd.get_dummies(weekly_model, columns=["Category", "season"], drop_first=True)

print(weekly_model.shape)
print(weekly_model.columns.tolist())
```

```python
feature_cols = [
    "Price", "Promotion",
    "Cost_Price", "Selling_Price", "Gross_Margin_Per_Unit",
    "is_holiday", "is_promo_event",
    "Current_Stock", "On_Order", "Lead_Time_Days", "Safety_Stock", "Reorder_Point", "Inventory_Value",
    "snapshot_gap_days",
    "units_lag_1", "units_lag_2", "units_lag_4", "units_roll_mean_4",
    "Category_Home Decor", "Category_Kitchen", "Category_Lighting", "Category_Storage",
    "season_Monsoon", "season_Spring", "season_Summer", "season_Winter"
]

X = weekly_model[feature_cols]
y = weekly_model["Units_Sold"]

print(X.shape, y.shape)
print(X.isnull().sum().sum())
```

```python
import numpy as np

def wape(actual, predicted):
    return np.abs(actual - predicted).sum() / actual.sum()

y_naive_pred = weekly_model["units_lag_1"]

naive_wape = wape(weekly_model["Units_Sold"], y_naive_pred)
print("Naive baseline WAPE:", naive_wape)
```

```python
from sklearn.ensemble import RandomForestRegressor

dates_sorted = sorted(weekly_model["Date"].unique())
n = len(dates_sorted)
print("Total weeks available:", n)

test_window = 8
n_folds = 3
fold_scores = []

for i in range(n_folds):
    cutoff = n - test_window * (n_folds - i)
    train_dates = dates_sorted[:cutoff]
    test_dates = dates_sorted[cutoff:cutoff + test_window]

    train_mask = weekly_model["Date"].isin(train_dates)
    test_mask = weekly_model["Date"].isin(test_dates)

    X_train, y_train = X[train_mask], y[train_mask]
    X_test, y_test = X[test_mask], y[test_mask]

    model = RandomForestRegressor(n_estimators=200, random_state=42)
    model.fit(X_train, y_train)
    preds = model.predict(X_test)

    score = wape(y_test, preds)
    fold_scores.append(score)
    print(f"Fold {i+1}: train through {train_dates[-1].date()}, "
          f"test {test_dates[0].date()} to {test_dates[-1].date()}, WAPE = {score:.4f}")

print("Average rolling-origin WAPE:", np.mean(fold_scores))
```

```python
print(
    weekly_model[weekly_model["Date"].isin(test_dates)]
    [["Date", "is_holiday", "is_promo_event"]]
    .drop_duplicates()
    .sort_values("Date")
)
```

```python
final_model = RandomForestRegressor(n_estimators=200, random_state=42)
final_model.fit(X, y)
print("Final model trained on", X.shape[0], "rows")
```

```python
latest = weekly.sort_values("Date").groupby("SKU").tail(1).copy()

print(latest.shape)
latest[["SKU", "Date", "Current_Stock", "Safety_Stock", "Reorder_Point",
        "Lead_Time_Days", "units_roll_mean_4"]].head()
```

```python
latest["weeks_of_cover"] = latest["Current_Stock"] / latest["units_roll_mean_4"]
latest["lead_time_weeks"] = latest["Lead_Time_Days"] / 7

latest["stockout_risk"] = (latest["Current_Stock"] <= latest["Reorder_Point"]).astype(int)
latest["overstock_risk"] = (latest["weeks_of_cover"] > 8).astype(int)

print(latest["stockout_risk"].sum(), "SKUs currently at/below reorder point")
print(latest["overstock_risk"].sum(), "SKUs holding more than 8 weeks of stock")

latest[["SKU", "Current_Stock", "Reorder_Point", "weeks_of_cover",
        "stockout_risk", "overstock_risk"]].sort_values("weeks_of_cover")
```

```python
latest["shortfall_weeks"] = (latest["lead_time_weeks"] - latest["weeks_of_cover"]).clip(lower=0)
latest["potential_lost_units"] = latest["shortfall_weeks"] * latest["units_roll_mean_4"]
latest["potential_lost_revenue"] = latest["potential_lost_units"] * latest["Selling_Price"]

latest["excess_units"] = (latest["Current_Stock"] - 8 * latest["units_roll_mean_4"]).clip(lower=0)
latest["excess_capital_tied_up"] = latest["excess_units"] * latest["Cost_Price"]

print("Total potential lost revenue (stockout risk):", latest["potential_lost_revenue"].sum())
print("Total capital tied up in excess stock:", latest["excess_capital_tied_up"].sum())
```
